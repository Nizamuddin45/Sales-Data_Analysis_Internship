```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("data/sales.csv")

# Compute Total Sales per order
df["Total"] = df["Quantity"] * df["Price"]

print("=================================================")

# 1. Overall Revenue
print("Total Revenue:", df["Total"].sum())

print("=================================================")


# 2. Average Order Value
print("Average Order Value:", df["Total"].mean())

print("=================================================")


# 3. Top Products
print("Top Products:\n", df.groupby("Product")["Total"].sum().sort_values(ascending=False).head(5))

print("=================================================")


# 4. Region-wise Sales
print("Sales by Region:\n", df.groupby("Region")["Total"].sum())

print("=================================================")


# 5. Category-wise Sales
print("Sales by Category:\n", df.groupby("Category")["Total"].sum())

print("=================================================")


# 6. Monthly Trend
df["Date"] = pd.to_datetime(df["Date"])
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total"].sum()
print("Monthly Sales:\n", monthly_sales)

print("=================================================")


# 7. Visualization
plt.figure(figsize=(8,5))
monthly_sales.plot(kind="bar", title="Monthly Sales Trend")
plt.show()
```
